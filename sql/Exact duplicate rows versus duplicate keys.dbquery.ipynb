```sql
SELECT CASE WHEN d._row_hash = s._row_hash THEN 'exact duplicate row'
            ELSE 'same key, different content' END AS duplicate_type,
       COUNT(*) AS row_count
FROM workspace.ecom_poc.duplicate_orders d
JOIN workspace.ecom_poc.silver_orders s ON d.order_id = s.order_id
GROUP BY 1;
```
