```sql
WITH sample_keys AS (SELECT DISTINCT order_id FROM workspace.ecom_poc.duplicate_orders LIMIT 5)
SELECT 'WINNER (kept)' AS outcome, s.order_id, s.quantity, s.order_status, s._row_hash
FROM workspace.ecom_poc.silver_orders s JOIN sample_keys k ON s.order_id = k.order_id
UNION ALL
SELECT 'LOSER (duplicate)', d.order_id, d.quantity, d.order_status, d._row_hash
FROM workspace.ecom_poc.duplicate_orders d JOIN sample_keys k ON d.order_id = k.order_id
ORDER BY order_id, outcome DESC;
```
