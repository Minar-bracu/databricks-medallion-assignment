```sql
SELECT order_id,
       quantity   AS raw_quantity,   try_cast(trim(quantity) AS INT)             AS quantity_after,
       unit_price AS raw_unit_price, try_cast(trim(unit_price) AS DECIMAL(10,2)) AS unit_price_after,
       order_date AS raw_order_date, try_to_date(trim(order_date), 'yyyy-MM-dd') AS order_date_after
FROM workspace.ecom_poc.bronze_orders
WHERE try_cast(trim(quantity) AS INT) IS NULL
   OR try_cast(trim(unit_price) AS DECIMAL(10,2)) IS NULL
   OR try_to_date(trim(order_date), 'yyyy-MM-dd') IS NULL
LIMIT 10;
```
